# Different Model integration

```
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
```

## Open AI v1

```
from langchain.chat_models import init_chat_model
model = init_chat_model(model="gpt-4.1")
response = model.invoke("how are you ? what is you fav color ?")
print(response)
```

```
print(response.content)
```

## OpenAI v2

```
from langchain_openai import OpenAI, ChatOpenAI

model = ChatOpenAI(model="gpt-4.1")
model
```

```
response = model.invoke("why india loose cwc23 ?")
print(response.content)
```

## Google Gemini Model Integration

```
from langchain.chat_models import init_chat_model
model = init_chat_model(model="google_genai:gemini-3.1-flash-lite")

response = model.invoke("why india loose cwc23 ?")
print(response.content)
```

```
print(response.content[0]["text"])
```

```
from langchain_google_genai import ChatGoogleGenerativeAI
model = ChatGoogleGenerativeAI(model="gemini-3.1-flash-lite")
response = model.invoke("Who is Ms Dhoni")
```

```
print(response.content[0]["text"])
```

## Groq Integration

```
import requests
import os

api_key = os.environ.get("GROQ_API_KEY")
url = "https://api.groq.com/openai/v1/models"

headers = {
    "Authorization": f"Bearer {api_key}",
    "Content-Type": "application/json"
}

response = requests.get(url, headers=headers)

print(response.json())
```

```
from langchain.chat_models import init_chat_model
model = init_chat_model(model="groq:openai/gpt-oss-120b")
response = model.invoke("Who is N. Modi")
```

```
print(response.content)
```

```
from langchain_groq import ChatGroq
model = ChatGroq(model="openai/gpt-oss-120b")

response = model.invoke("Hello, I am Rahul Gandhi, Tell me about my grandfather ?")
```

```
print(response.content)
```
